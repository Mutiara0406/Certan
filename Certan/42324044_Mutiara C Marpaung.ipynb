```python
# Perbedaan array dan list
import numpy as np

a = np.array([1, 2, 3, 4])
b = [1, 2, 3, 4]

print(a, b)
```

```python
# operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a + b
hasil2 = a - b
hasil3 = a * b
hasil4 = a / b
hasil5 = a**2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
# operator Elementwise ada masalah di perkalian matriks karena mengalikan per elemen
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[3, 4, 5], [3, 4, 5]])

hasil = a * b
print("hasil =", hasil)
```

```python
# perkalian matriks
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[1, 2, 3], [1, 2, 3], [1, 2, 3]])

hasil_perkalian_matriks = np.dot(a, b)
print(hasil_perkalian_matriks)
```

```python
# indexing
a = np.arange(10)**3
print(a)

print("mengambil elemen ke 1 dari a =", a[0])
print("mengambil elemen ke 4 dari a =", a[3])
print("mengambil elemen terakhir dari a =", a[-1])
print("mengambil elemen ke 1 dari a =", a[0])
print("mengambil elemen dari 1 sampai 4 =", a[0:4])
print("mengambil elemen dari 4 sampai terakhir =", a[3:])
print("mengambil elemen dari awal sampai 4 =", a[:4])
```

```python
# mengetahui ukuran matriks
a = np.array([[1, 2, 3], [4, 5, 6]])
print("matriks a berukuran =", a.shape)

# transpose matriks
print("transpose matriks a =", a.transpose())
print("transpose matriks a =", np.transpose(a))
print("transpose matriks a =", a.T)

# Pada Machine Learning biasanya flatten dan stacking sangat penting
# flatten array, vektor baris
print("flatten dari matriks a =", a.ravel())
```

```python
# Pada Machine learning biasanya flatten dan stacking sangat penting
# stacking array atau menumpuk array
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

a1 = np.hstack((a, b))
a2 = np.vstack((a, b))

print("horizontal stack =", a1)
print("vertikal stack =", a2)
```

```python
# invers matriks
a = np.array([[1, -1], [1, 1]])
print(a)

a_inv = np.linalg.inv(a)
print(a_inv)

det_a = np.linalg.det(a)
print(det_a)
```

```python
# solusi persamaan linear
A = np.array([[1, -6], [2, -5]])
Y = np.array([-35, 21])
print(A)
print(Y)

A_inv = np.linalg.inv(A)

# menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

# cara lain
X2 = np.linalg.solve(A, Y)
print(X2)
```

```python
def f(x, y):
    return np.cos(np.exp(-x**2 - y**2)) * np.sin(np.sqrt(x**2 + y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat grid x dan y
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# Persamaan fungsi 2D
F = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2 - 1

# Plotting grafik 2D menggunakan contour (di mana F = 0)
plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, [0], colors='red')
plt.title("Grafik 2D: $x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1$")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# Definisi fungsi Z
def f(x, y):
    return ((5*y / 4) - np.abs(x))**2

x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Plotting 3D
fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

ax.set_title(r'Grafik 3D: $z = \left(\frac{5y}{4} - |x|\right)^2$')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
plt.show()
```
