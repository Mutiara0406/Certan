Simple Linear Regression

```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=0
)
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
y_pred = regressor.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
score
```

```python
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Multiple Linear Regression

```python
df2 = pd.read_csv('multiple_linear_regression_dataset.csv')
print(df2.head())
```

```python
X2 = df2.iloc[:, :-1]
y2 = df2.iloc[:, 2]
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X2, y2, test_size=0.2, random_state=0
)
```

```python
regressor = LinearRegression()
regressor.fit(X_train2, y_train2)
```

```python
y_pred2 = regressor.predict(X_test2)
print(y_pred2)
```

```python
a_ = regressor.intercept_
b1 = regressor.coef_[0]
b2 = regressor.coef_[1]
print("Estimated model intercept, a:", a_)
print("Estimated model slope, b1:", b1)
print("Estimated model slope, b2:", b2)
print(f"ŷ = {a:.2f} {b1:.2f}x1 + {b2:.2f}x2")
```

```python
score2=r2_score(y_test2,y_pred2)
score2
```

```python
import pandas as pd

df = pd.read_csv('Student_Scores.csv')

print(df.head())
```

```python
print(df.columns)
```

```python
X = df[['Hours']].values
y = df['Scores'].values
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='red')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.title('Study Hours vs Score')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()

regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)

print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
y_pred = regressor.predict(X_test)

print("Hasil prediksi:")
print(y_pred)
```

```python
from sklearn.metrics import r2_score

score = r2_score(y_test, y_pred)

print("R-Square:", score)
```

```python
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')

plt.title('Study Hours vs Score (Training Set)')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.show()
```

```python
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")

print("Hasil prediksi:")
print(y_pred)

print("R-Square:", score)
```

```python
import pandas as pd

df2 = pd.read_csv('Marketing_Data.csv')

print(df2.head())
```

```python
X2 = df2[['youtube', 'facebook', 'newspaper']].values
y2 = df2['sales'].values
```

```python
from sklearn.model_selection import train_test_split

X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X2, y2, test_size=0.2, random_state=0
)
```

```python
from sklearn.linear_model import LinearRegression

regressor2 = LinearRegression()
regressor2.fit(X_train2, y_train2)
```

```python
y_pred2 = regressor2.predict(X_test2)

print("Hasil prediksi:")
print(y_pred2)
```

```python
a2 = regressor2.intercept_
b1 = regressor2.coef_[0]
b2 = regressor2.coef_[1]
b3 = regressor2.coef_[2]

print("Estimated model intercept, a:", a2)
print("Estimated model coefficient, b1 (youtube):", b1)
print("Estimated model coefficient, b2 (facebook):", b2)
print("Estimated model coefficient, b3 (newspaper):", b3)

print(f"ŷ = {a2:.2f} + {b1:.2f}x1 + {b2:.2f}x2 + {b3:.2f}x3")
```

```python
from sklearn.metrics import r2_score

score2 = r2_score(y_test2, y_pred2)

print("R-Square:", score2)
```
