```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
```

numpy digunakan untuk membantu proses pengolahan data dalam bentuk array, sedangkan pandas digunakan untuk mengolah data dalam bentuk tabel atau DataFrame. Setelah library disiapkan, dataset Iris dimuat menggunakan fungsi load_iris() dari scikit-learn. Dataset Iris digunakan sebagai data yang akan diproses dan diklasifikasikan menggunakan algoritma Naïve Bayes.

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Data dari dataset Iris diubah menjadi bentuk DataFrame menggunakan pandas agar lebih mudah ditampilkan dan diolah. iris.data berisi nilai dari empat fitur bunga, sedangkan iris.feature_names digunakan untuk memberikan nama pada masing-masing kolom. Setelah itu, kolom Species ditambahkan untuk menunjukkan jenis bunga berdasarkan target yang terdapat dalam dataset.

Data dipisahkan menjadi bagian fitur dan target. Variabel X berisi empat fitur yang digunakan untuk menentukan jenis bunga, yaitu panjang sepal, lebar sepal, panjang petal, dan lebar petal. Kolom Species dihapus dari X karena kolom tersebut merupakan hasil yang akan diprediksi. Sementara itu, y berisi jenis species yang menjadi target atau label. Perintah print(X) digunakan untuk menampilkan data fitur yang telah dipisahkan.

```python
y
```

Variabel y digunakan untuk menyimpan target atau label dari dataset Iris. Data yang ditampilkan berupa nama species dari setiap bunga, yaitu setosa, versicolor, dan virginica. Target ini nantinya akan digunakan sebagai hasil yang harus dipelajari oleh model dalam proses klasifikasi.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

LabelEncoder digunakan untuk mengubah label species yang awalnya berupa teks menjadi angka. Pengubahan ini diperlukan agar label dapat diproses oleh model machine learning. Setelah dilakukan encoding, setosa, versicolor, dan virginica masing-masing direpresentasikan dengan angka yang berbeda.

Hasil dari proses LabelEncoder ditampilkan dalam bentuk array angka. Nilai angka tersebut menjadi representasi dari masing-masing species. Dengan demikian, data target yang sebelumnya berupa teks sudah dapat digunakan dalam proses pelatihan model Naïve Bayes.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Dataset dibagi menjadi data training dan data testing. Data training digunakan untuk melatih model agar dapat mempelajari pola dari dataset, sedangkan data testing digunakan untuk menguji hasil prediksi model. Nilai test_size=0.2 berarti 20% data digunakan sebagai data testing. random_state=0 digunakan agar pembagian data menghasilkan susunan yang sama ketika program dijalankan kembali.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

GaussianNB digunakan untuk menerapkan algoritma Gaussian Naïve Bayes. Jenis Naïve Bayes ini digunakan karena fitur pada dataset Iris berupa data numerik kontinu. Empat fitur yang digunakan, yaitu panjang dan lebar sepal serta panjang dan lebar petal, merupakan data berupa angka. Setelah GaussianNB() dibuat, variabel model akan digunakan untuk proses pelatihan dan prediksi.

Model dilatih menggunakan data training yang terdiri dari X_train sebagai fitur dan y_train sebagai target. Proses fit() membuat model mempelajari pola hubungan antara nilai fitur bunga dengan species yang sesuai. Model yang sudah selesai dilatih disimpan dalam variabel NB_model dan nantinya digunakan untuk melakukan prediksi.

```python
result = NB_model.predict(X_test)
result
```

Model yang sudah dilatih digunakan untuk memprediksi species dari data testing. X_test diberikan sebagai input ke fungsi predict(), kemudian hasil prediksi disimpan dalam variabel result. Hasil tersebut berupa angka yang mewakili masing-masing species karena sebelumnya label telah diubah menggunakan LabelEncoder.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

accuracy_score() digunakan untuk mengetahui seberapa banyak hasil prediksi yang sesuai dengan data sebenarnya. Hasil prediksi pada result dibandingkan dengan target sebenarnya yang terdapat pada y_test. Nilai akurasi kemudian dikalikan dengan 100 agar ditampilkan dalam bentuk persentase. Berdasarkan modul, hasil akurasi yang diperoleh adalah 96,67%.

```python
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Data baru disiapkan untuk melihat kemampuan model dalam melakukan prediksi terhadap data yang belum pernah digunakan sebelumnya. Data tersebut memiliki empat nilai yang sesuai dengan empat fitur pada dataset Iris. Nilai tersebut kemudian disimpan dalam variabel input_data untuk digunakan sebagai masukan ke model.

Data baru yang telah disiapkan dimasukkan ke model menggunakan fungsi predict(). Model kemudian menggunakan pola yang telah dipelajari dari data training untuk menentukan species yang paling sesuai dengan karakteristik data tersebut. Hasil prediksi disimpan dalam variabel result.

Hasil prediksi dari model masih berupa angka karena sebelumnya label species telah diubah menggunakan LabelEncoder. Fungsi inverse_transform() digunakan untuk mengembalikan angka tersebut menjadi nama species sehingga hasil prediksi lebih mudah dipahami.

Perintah print() digunakan untuk menampilkan hasil prediksi dalam bentuk nama species. Berdasarkan contoh yang terdapat pada modul, data dengan nilai [5.1, 3.5, 1.4, 0.2] menghasilkan prediksi setosa.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Data input dibuat dalam bentuk DataFrame dan diberikan nama kolom yang sama dengan nama fitur pada dataset Iris. Hal ini dilakukan agar struktur data yang digunakan saat prediksi sama dengan struktur data ketika model dilatih. Pada contoh sebelumnya muncul peringatan karena data input tidak memiliki nama fitur, sehingga penggunaan DataFrame dengan nama kolom dapat menghindari perbedaan tersebut.

Data baru yang sudah dibuat dalam bentuk DataFrame digunakan kembali sebagai input untuk model. Model kemudian memproses nilai dari keempat fitur tersebut dan menghasilkan prediksi species berdasarkan pola yang telah dipelajari selama proses training.

Hasil prediksi yang masih berbentuk angka diubah kembali menjadi nama species menggunakan inverse_transform(). Dengan begitu, hasil prediksi dapat langsung diketahui dalam bentuk nama species seperti setosa, versicolor, atau virginica.

Perintah ini digunakan untuk menampilkan hasil akhir prediksi dari data baru. Dengan adanya perintah ini, hasil prediksi dapat dilihat dalam bentuk nama species, bukan lagi dalam bentuk angka. Pada contoh di modul, hasilnya adalah setosa.
