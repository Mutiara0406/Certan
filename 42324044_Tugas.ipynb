```python
import numpy as np

# Membuat matriks 5x5 dengan nilai acak (0 sampai 100)
matriks = np.random.randint(0, 100, size=(5, 5))

# Menghitung nilai maksimum, minimum, dan rata-rata
nilai_max = matriks.max()
nilai_min = matriks.min()
nilai_mean = matriks.mean()

print("Matriks 5x5:")
print(matriks)
print("\nNilai Maksimum :", nilai_max)
print("Nilai Minimum  :", nilai_min)
print("Rata-rata      :", nilai_mean)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat grid x dan y
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# Persamaan fungsi 2D
F = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2 - 1

# Plotting grafik 2D menggunakan contour (di mana F = 0)
plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, [0], colors='red')
plt.title("Grafik 2D: $x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1$")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# Definisi fungsi Z
def f(x, y):
    return ((5*y / 4) - np.abs(x))**2

x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Plotting 3D
fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

ax.set_title(r'Grafik 3D: $z = \left(\frac{5y}{4} - |x|\right)^2$')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)
plt.show()
```
